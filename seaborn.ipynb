```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
np.random.seed(42)
scores = np.random.normal(loc = 70, scale = 10, size = 200)

#plot the histogram
plt.figure(figsize=(10,6))
sns.histplot(scores, bins = 20, kde = True)


#draw the vertical lime for the mean
mean_score = np.mean(scores)
plt.axvline(mean_score, color = 'red', linestyle = '--', label = f'Mena: {mean_score:.2f}')

#labels and title
plt.title('Distribution of Exam Scores',fontsize = 16)
plt.xlabel('Score')
plt.ylabel('Count')
plt.legend()
plt.annotate(f'Mean: {mean_score:.2f}', xy = (mean_score, 5), xytext = (mean_score + 5, 10), arrowprops = dict(facecolor = 'red', shrink = 0.05))
plt.show()  
```

```python
#load dataset
penguins = sns.load_dataset('penguins')

#plot
plt.figure(figsize=(10,6))
sns.scatterplot(
    data = penguins,
    x = 'flipper_length_mm',
    y = 'body_mass_g',
    hue = 'species',
    palette = 'Set2',
)

plt.title('Flipper Length vs. Body Mass by Species', fontsize = 14)
plt.xlabel('Flipper Length (mm)')
plt.ylabel('Body Mass (g)')
plt.show()
```
